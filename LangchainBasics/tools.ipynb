This @tool decorator is the simplest way to define a custom tool. The decorator uses the function name as the tool name by ddefault , but this can be overridden by passing a string as the first arguement . Additionally , the decorator will use the function's docstring as the tool's description - so a docstring must be provided

```python
##tool decorator
from langchain_core.tools import tool
@tool
def division(a:int,b:int)->int:
    """"Divide 2 number"""
    return a/b
```

```python
print(division.name)
print(division.description)
print(division.args)
```

```python
## async implementation
##tool decorator
from langchain_core.tools import tool
@tool
async def multiply(a:int,b:int)->int:
    """"Multiply2 number"""
    return a*b
```

```python
from typing import Annotated,List
@tool
def multiply_by_max(
    a:Annotated[int , "A value"],
    b:Annotated[List[int] , "b value"]
)->int:
    """Multiply a by the maximum of b"""
    return a*max(b)
```

```python
print(multiply_by_max.args)
```

<h5>
<p>
Structured Tool
The StructuredTool.from_function class method provides a  but more configurability that the @too; decorator , without requiring much additional code.
</p>
</h5>

```python
from langchain_core.tools import StructuredTool


def multiply(a:int,b:int)->int:
    """"Multiply two number"""
    return a * b


async def amultiply(a:int,b:int)->int:
    """"Multiply two number"""
    return a * b

calculator= StructuredTool.from_function(func=multiply , coroutine=amultiply)

print(calculator.invoke({"a":2 , "b":3}))
print(await calculator.ainvoke({"a":2 , "b":5}))

## async -> used for asynchronous execution  
```

<h3>Call Tool with LLM models</h3>

```python

import wikipedia
import inspect

# print(wikipedia.__version__)
# print(inspect.signature(wikipedia.set_user_agent))

wikipedia.set_user_agent(
    "AgenticAI-Learning/1.0 (raj.akshat2202@gmail.com)"
)

from langchain_community.tools import WikipediaQueryRun
from langchain_community.utilities import WikipediaAPIWrapper

api_wrapper = WikipediaAPIWrapper(
    top_k_results=5,
    doc_content_chars_max=1500
)

wiki_tool = WikipediaQueryRun(api_wrapper=api_wrapper)

print(wiki_tool.invoke({"query": "Robert Downey Junior"}))
```

```python

from langchain_core.tools import tool


@tool
def add(a: int, b: int) -> int:
    """Adds a and b."""
    return a + b


@tool
def multiply(a: int, b: int) -> int:
    """Multiplies a and b."""
    return a * b
```

```python
from langchain_tavily import TavilySearch

tavily_tool = TavilySearch(
    max_results=5,
    topic="general",
    # include_answer=False,
    # include_raw_content=False,
    # include_images=False,
    # include_image_descriptions=False,
    # search_depth="basic",
    # time_range="day",
    # include_domains=None,
    # exclude_domains=None
)
```

```python
tools=[wiki_tool,add,multiply,tavily_tool]
```

```python
tools
```

```python

from langchain.chat_models import init_chat_model
llm=init_chat_model("qwen/qwen3.6-27b",model_provider="groq")
llm
```

```python
llm_with_tools = llm.bind_tools([
    add,
    multiply,
    wiki_tool,
    tavily_tool
])
llm_with_tools
```

```python
from langchain_core.messages import HumanMessage
query="What is 2* 3"
messages = [HumanMessage(query)]
response=llm_with_tools.invoke(messages)
messages.append(response)

print(response)
```

```python

response.tool_calls
```

```python
from langchain_core.messages import HumanMessage

query = "What is 2*3"
messages = [HumanMessage(query)]

# Step 1: LLM decides to call a tool
response = llm_with_tools.invoke(messages)
messages.append(response)          # <-- YEH LINE MISSING THI, add karo

# Step 2: Actually run the tool
for tool_call in response.tool_calls:
    selected_tool = {
        "add": add,
        "multiply": multiply,
        "wikipedia": wiki_tool,
        "tavily_search": tavily_tool
    }[tool_call["name"].lower()]
    tool_msg = selected_tool.invoke(tool_call)
    messages.append(tool_msg)

# Step 3: Feed everything back — ab yahan real answer aayega
final_response = llm_with_tools.invoke(messages)
print(final_response.content)
```
